```python
12 & 43    #  001100   → 12
           # & 101011   → 43
            #   ----------
           #   001000   → 8
```

```python
12 | 15      #     1100   → 12
             #   | 1111   → 15
             #      --------
             #     1111   → 15
```

```python
12 ^ 13    #    1100   → 12
           #  ^ 1101   → 13
           #     --------
           #    0001   → 1
```

```python
12 << 2        # Easy rule

                #   For left shift:

                #        number << n = number × 2ⁿ

                #   So:

                 #     12 × 2² = 12 × 4 = 48.
```

```python
12>>1         # Easy rule

               # Right shift by n means approximately:

                #   number ÷ 2ⁿ

              #  So:

              #     12 ÷ 2¹ = 6.
```
