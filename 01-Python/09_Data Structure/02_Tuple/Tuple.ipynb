# Tuple Methods

A tuple has `2` built-in methods: `count()` and `index()`

### 1. count()

Used to count how many times a specified value occurs in a tuple.

Syntax:  tuple.`count`(value)

```python
a = (10, 20, 10, 30, 10, 40)

print(a.count(10))
```

### 2.index()

Used to find the index of the first occurrence of a value.

```python
a = (10, 20, 30, 40)

print(a.index(30))
```

## Adding elements

You cannot directly add or modify elements in an existing tuple because tuples are immutable.

However, you can create a new tuple by using `concatenation`.

```python
a = (1, 2, 3)
```

```python
a = a + (5,)
```
