## Looping and Built-in Functions with Lists

### 1. Using for Loop with a List

A for loop can be used to access each element of a list one by one.

```python
a = [1,2,3,4,5]
```

```python
for i in a:
    print(i)
```

### 2. Using for i in range()

We use for i in range(len(list)) to access list elements using their index.

```python
a = [10, 20, 30, 40, 50]

for i in range(len(a)):
    print(a[i])
```

How it works:

len(a) → gives the number of elements → 5

range(5) → generates 0, 1, 2, 3, 4

i → gets each index

a[i] → gets the element at that index

### 3. Using enumerate()

enumerate() is used to get both the index and the element from a list.

```python
a = [10, 20, 30, 40]

for i, value in enumerate(a):
    print(i, value)
```

Simple explanation: 

i → stores the index

value → stores the element

enumerate(a) → gives index and element together


### 4. all()

`all()` returns `True` only when all elements are truthy; otherwise, it returns `False`.

```python
a = [True, True, True]

print(all(a))
```

```python
a = [True, False, True]

print(all(a))
```

```python
a = [1, 2, 3]                 # Because all numbers are non-zero (truthy).
print(all(a))
```

```python
a = [1, 0, 3]                # Because 0 is considered False.
print(all(a))   
```

### 5. any()

`any()` checks whether at least one element is `True` (truthy).

```python
a = [False, False, True]         # Because one element is True.

print(any(a))
```

```python
any(a)
```

```python
a = [False, False, False]

print(any(a))
```

```python
a = []                # An empty list has no `True` (truthy) element, so `any()` returns `False`.

print(any(a))
```

Remember:

`any([])` → `False`

`all([])` → `True`

### Python List — All Important Points

1. **List** is an ordered and mutable collection of elements.
2. Lists are represented using **square brackets `[]`**.
3. Lists allow **duplicate elements**.
4. Lists can contain **different data types**.
5. Lists support **positive and negative indexing**.
6. Lists support **slicing**.
7. Lists are **mutable**, so elements can be changed.
8. `len()` returns the **number of elements** in a list.
9. `append()` adds an element to the **end** of the list.
10. `insert()` adds an element at a **specific index**.
11. `extend()` adds **multiple elements** to the list.
12. `remove()` removes the **first occurrence** of a specified value.
13. `pop()` removes and returns an element using its **index**.
14. `clear()` removes **all elements** from the list.
15. `del` can delete an element, a range of elements, or the entire list.
16. `sort()` sorts the list in **ascending order by default**.
17. `sort(reverse=True)` sorts the list in **descending order**.
18. `reverse()` simply **reverses the order** of elements.
19. `sorted()` returns a **new sorted list** without changing the original list.
20. `count()` returns the **number of occurrences** of a value.
21. `index()` returns the **index of the first occurrence** of a value.
22. `in` checks whether an element **exists** in the list.
23. `not in` checks whether an element **does not exist** in the list.
24. `+` is used to **concatenate** two lists.
25. `*` is used to **repeat** the elements of a list.
26. `min()` returns the **smallest** element.
27. `max()` returns the **largest** element.
28. `sum()` returns the **total of numeric elements**.
29. `all()` returns `True` when **all elements are truthy**.
30. `any()` returns `True` when **at least one element is truthy**.
31. `for` loop is used to **iterate through list elements**.
32. `for i in range()` is used to access list elements using their **index**.
33. `enumerate()` provides both the **index and value** while iterating.
34. **List unpacking** assigns list elements to multiple variables.
35. Lists can contain **nested lists**.
36. **List comprehension** provides a concise way to create lists.
37. `copy()` creates a **copy of the list**.
38. Lists are **dynamic**, so their size can change during execution.
