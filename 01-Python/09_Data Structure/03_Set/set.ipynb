# Set

### A set is an unordered, mutable collection of unique elements in Python.

##### -> Sets do not allow duplicate values.

##### -> Sets are unordered, so they do not maintain a fixed index.

##### -> Sets are mutable, so elements can be added or removed.

##### -> Sets can contain different data types.

##### -> Sets are created using curly braces {} with values.

##### -> An empty {} creates a dictionary, not a set.

##### -> An empty set is created using set().

```python
s = {}
```

```python
print(type(s))
```

```python
s1 = {10,202,40}
print(type(s1))
```

```python
s = set()
print(type(s))
```

```python
print(s)
print(s1)
```

# Set Operations

## 1. union

Combines the elements of two or more sets and removes duplicate elements.

```python
a = {1,2,3,4,5}
b = {4,5,6,7,8}
c = {8,9,10} 
```

```python
a.union(b)
```

```python
b.union(a,c)
```

```python
a | b | c
```

```python
print(a)
print(b)
print(c)
```

## 2. intersection

Returns the common elements present in two or more sets.

```python
a = {1,2,3,4,5}
b = {4,5,6,7,8}
c = {8,9,10} 
```

```python
a.intersection(b)
```

```python
b.intersection(c)
```

```python
b & c
```

```python
a & c
```

```python
a & b
```

```python
a & b & c
```

## 3. difference()

Returns the elements that are present in the first set but not in the second set.

```python
a = {1,2,3,4,5}
b = {4,5,6,7,8}
c = {8,9,10} 
```

```python
a.difference(b)
```

```python
b.difference(c)
```

```python
a - b
```

```python
b - c
```

```python
c - a
```

```python
a - c
```

## 4. symmetric_difference

Returns the elements that are not common between two sets.

```python
a = {1,2,3,4,5}
b = {4,5,6,7,8}
c = {8,9,10} 
```

```python
a.symmetric_difference(b)
```

```python
a ^ b
```

```python
b ^ c
```

## 5. symmetric_difference_update()

Updates the first set with non-common elements from both sets.

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

a.symmetric_difference_update(b)

print(a)
```

# Set Comparison Methods

# 1. issubset()

Checks whether all elements of one set are present in another set.

```python
a = {1, 2}
b = {1, 2, 3, 4}

print(a.issubset(b))
```

## 2. issuperset()

Checks whether a set contains all elements of another set.

```python
a = {1, 2, 3, 4}
b = {1, 2}

print(a.issuperset(b))
```

## 3. isdisjoint()

Checks whether two sets have no common elements.

```python
a = {1, 2}
b = {3, 4}

print(a.isdisjoint(b))
```

# Python Set — All Important Concepts

 ## 1. Definition
 
A set is an unordered collection of unique elements.
Sets are written using {}.
Sets automatically remove duplicate elements.

## 2. Properties of Set

Allows unique elements only.
Does not allow duplicate elements.
Set elements must be hashable.
Can contain different data types if they are hashable.
Sets are mutable.
Sets do not support indexing.
Sets do not support slicing.
Sets are unordered.

## 3. Creating a Set
Set using {}.
Empty set using set().
{} creates an empty dictionary, not an empty set.

## 4. Adding Elements

add() → Adds one element to a set.

update() → Adds multiple elements to a set.

## 5. Removing Elements

remove() → Removes a specified element; gives an error if the element is not found.

discard() → Removes a specified element; does not give an error if it is not found.

pop() → Removes one element from the set.

clear() → Removes all elements from the set.

## 6. Set Operations

union() → Combines elements from sets.

intersection() → Returns common elements.

difference() → Returns elements only from the first set.

symmetric_difference() → Returns non-common elements from both sets.

## 7. Set Update Operations

update() → Adds elements from another collection.

intersection_update() → Keeps only common elements.

difference_update() → Removes elements found in another set.

symmetric_difference_update() → Keeps only non-common elements.

## 8. Set Comparison Methods

issubset() → Checks whether all elements of one set are present in another set.

issuperset() → Checks whether a set contains all elements of another set.

isdisjoint() → Checks whether two sets have no common elements.

## 9. Other Important Methods

copy() → Creates a copy of the set.

len() → Finds the number of elements.

in → Checks whether an element exists in the set.

not in → Checks whether an element does not exist in the set.

## 10. Set Operators

| → Union

& → Intersection
- → Difference
  
^ → Symmetric difference

<= → Subset

>= → Superset
