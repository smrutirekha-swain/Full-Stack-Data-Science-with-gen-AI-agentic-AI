###  Math Module 

```python
# math is a built-in Python module that provides mathematical functions and constants.
```

```python
import math
```

```python
print(math.sqrt(25))      # 5.0
print(math.pow(2, 3))     # 8.0
print(math.factorial(5))  # 120
print(math.ceil(4.2))     # 5
print(math.floor(4.8))    # 4
```

### sqrt()

```python
x = math.sqrt(25)    # Used to find the square root of a number.
x
```

### floor()

```python
print(math.floor(2.9)) #floor - minimum or least value 
```

```python
print(math.floor(4.8))    #Returns the largest integer less than or equal to the given number.
print(math.floor(7.9))
print(math.floor(-4.2))
```

### pow()  

```python
 # Used to calculate the power of a number.
```

```python
print(math.pow(2, 3))
print(math.pow(5, 2))
print(math.pow(10, 3))
```

### ceil()

```python
print(math.ceil(4.2))
print(math.ceil(7.1))
print(math.ceil(-4.2))
```

### factorial()

```python
print(math.factorial(5))
print(math.factorial(6))
print(math.factorial(0))
```

```python
# Used to find the factorial of a non-negative integer.
```

```python
# Example  :   5! = 5 × 4 × 3 × 2 × 1 = 120
```
