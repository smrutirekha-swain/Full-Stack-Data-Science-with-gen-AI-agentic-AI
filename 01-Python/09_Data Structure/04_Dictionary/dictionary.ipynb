# Dictionary

A dictionary is a mutable collection of key-value pairs in Python.

Dictionary = Collection of key-value pairs.

### Main Points :
 -> Stores data in key-value pairs.
 
 -> Each key must be unique.
 
 -> Values can be duplicated.
 
 -> Dictionary is mutable.
 
 -> Supports different data types.
 
 -> Values are accessed using their keys.
 
 -> Keys must be immutable/hashable.
 
 -> Dictionaries maintain insertion order.
 
 -> Created using {} or dict().

#### Syntax :
        dictionary = {
                      key: value
                      }

```python
student = {
    "name": "Rekha",
    "age": 25,
    "course": "Python"
}
```

```python
student
```

# Dictionary Methods

### 1. get()

Used to get the value of a specified key.

```python
print(student.get("name"))
```

### 2. keys()

Used to get all the keys.

```python
print(student.keys())
```

### 3. values()

Used to get all the values.

```python
print(student.values())
```

### 4. items()

Used to get all key-value pairs.

```python
print(student.items())
```

### 5. update()

Used to add new key-value pairs or update existing values.

```python
student.update({"age": 24})
print(student)
```

### 6. pop()

Used to remove a specified key-value pair.

```python
student.pop("age")
print(student)
```

### 7. popitem()

Used to remove the last inserted key-value pair.

```python
student.popitem()
print(student)
```

### 8. clear()

Used to remove all elements from the dictionary.

```python
student.clear()
print(student)
```

### 9. copy()

Used to create a copy of a dictionary.

```python
student = {"name" : "smruti"}
```

```python
student2 = student.copy()
print(student2)
```

### 10. setdefault()

Returns the value of a key. If the key does not exist, it adds the key with the specified value.

```python
student.setdefault("city", "Delhi")
print(student)
```

### 11. fromkeys()

Used to create a new dictionary using specified keys.

```python
keys = ["name", "age", "course"]

student = dict.fromkeys(keys)
print(student)
```

```python
keys = ["name":"smruti", "age": "23", "course":"Data science"]

student = dict.fromkeys(keys)
print(student)
```

fromkeys() is used to create a dictionary from a collection of keys, with one common value for all keys.

```python
keys = ["name", "age", "course"]

student = dict.fromkeys(keys, "srs")
print(student)
```
